```python
import duckdb
import os
```

```python
events_path = os.path.expanduser("~/projects/rolefit/data/processed/match_3754217_events.parquet")

query = f"SELECT * FROM read_parquet('{events_path}') LIMIT 5"
duckdb.sql(query).show()
```

```python
minutes_path = os.path.expanduser("~/projects/rolefit/data/processed/match_3754217_minutes.parquet")

minutes_query = f"SELECT * FROM read_parquet('{minutes_path}') LIMIT 5"
duckdb.sql(minutes_query).show()
```

```python
total_events_player = f"""With total_a as(
Select player, count(player) as total_events
From read_parquet('{events_path}')
Group by player)
,total_b as(
Select player, sum(minutes_played) as total_mins
From read_parquet('{minutes_path}')
Group by player)
Select total_a.player, total_events, total_mins
From total_a inner join total_b
On total_a.player = total_b.player"""

duckdb.sql(total_events_player).show()
```
