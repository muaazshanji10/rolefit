```python
import pandas as pd

csv_text = """player,goals,team,match_date
Messi,30,Inter Miami,2024-03-01
ronaldo ,28,Al Nassr,2024/03/05
Mbappe,twenty five,Real Madrid,05-03-2024
Messi,30,Inter Miami,2024-03-01
Haaland,32,Man City,2024-03-10
,29,PSG,2024-03-12
"""

with open("dirty_players.csv", "w") as f:
    f.write(csv_text)

df_dirty2 = pd.read_csv("dirty_players.csv")
print(df_dirty2)
print(df_dirty2.dtypes)
```

```python
print(df_dirty2.duplicated())
df_dirty2dupes = df_dirty2.drop_duplicates()
print(df_dirty2dupes)
print()
df_dirty3 = df_dirty2dupes.fillna("Unknown")
df_dirty3["player"] = df_dirty3["player"].str.strip()

print()
df_dirty3["player"] = df_dirty3["player"].str.title()


df_dirty3["goals"] = pd.to_numeric(df_dirty3["goals"], errors="coerce")
print(df_dirty3)
print()
df_dirty3["match_date"] = pd.to_datetime(df_dirty3["match_date"], format="mixed")
print(df_dirty3)
```
