```python
import pandas as pd
import numpy as np

data = {
    "name": ["Alice", "bob", "Charlie ", "Alice", None],
    "age": ["25", "30", "twenty-two", "25", "40"],
    "signup_date": ["2023-01-15", "2023/02/20", "15-03-2023", "2023-01-15", "2023-04-10"]
}
df_dirty = pd.DataFrame(data)
print(df_dirty)
print(df_dirty.dtypes)
```

```python
print(df_dirty.duplicated())
df_clean = df_dirty.drop_duplicates()
print(df_clean)
```

```python
df_clean["name"] = df_clean["name"].fillna("Unknown")
print(df_clean)
```

```python
df_clean["name"] = df_clean["name"].str.strip()
print(df_clean["name"])
```

```python
df_clean["name"] = df_clean["name"].str.title()
print(df_clean["name"])
```

```python
df_clean["age"] = pd.to_numeric(df_clean["age"], errors="coerce")
print(df_clean["age"])
print(df_clean["age"].dtype)
```

```python
#df_clean["signup_date"] = pd.to_datetime(df_clean["signup_date"])
#print(df_clean["signup_date"])
#print(df_clean["signup_date"].dtype) mixed format confused python

df_clean["signup_date"] = pd.to_datetime(df_clean["signup_date"], format="mixed")
print(df_clean["signup_date"])
print(df_clean["signup_date"].dtype)

```
