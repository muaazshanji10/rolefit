```python
import pandas as pd

data = {
    "player": ["Messi", "Ronaldo", "Mbappe", "Haaland"],
    "goals": [30, 28, 25, 32],
    "team": ["Inter Miami", "Al Nassr", "Real Madrid", "Man City"]
}
df = pd.DataFrame(data)
print(df)

print(df["goals"])          # shows as a Series, with an index
print(df["goals"].values)   # shows as a plain NumPy array, no labels

print(df.loc[2])      # Mbappe's row, by label
print(df.iloc[2])     # third row, by position
```

```python
df_sorted = df.sort_values("goals")
print(df_sorted)

print(df_sorted.loc[2])
print(df_sorted.iloc[0])
```

So loc gives you the row and normal printing gives you the column

```python
print(df[df["goals"] > 28])

print(df["goals"] > 28)

df["bonus"] = 10
print(df)
```

```python
high_scorers = df[df["goals"] > 28]
high_scorers["goals"] = high_scorers["goals"] + 100
print(high_scorers)
print(df)
```

```python
high_scorers = df[df["goals"] > 28].copy()
high_scorers["goals"] = high_scorers["goals"] + 100
print(high_scorers)
print(df)
```

```python
import pandas as pd

df_match = pd.read_json("/Users/muaazshanji/projects/statsbomb-data/data/events/15946.json")
print(df_match.shape)
print(df_match.columns.tolist())
```

```python
print(df_match["type"].iloc[0])
print(df_match["player"].iloc[0])
```

```python
df_match["type_name"] = df_match["type"].apply(lambda x: x["name"])
print(df_match["type_name"].value_counts())
```

```python
df_match["player_name"] = df_match["player"].apply(lambda x: x["name"] if pd.notna(x) else None)
print(df_match["player_name"].value_counts())
```

```python
print(df_match[["type_name", "player_name"]].head(10))
```

```python
passes_only = df_match[df_match["type_name"] == "Pass"]
print(passes_only["player_name"].value_counts())
```

```python
df_match["pass_outcome"] = df_match["pass"].apply(lambda x: x["outcome"]["name"] if pd.notna(x) and x.get("outcome") else None)
print(df_match["pass_outcome"].value_counts())
```

```python
df_match["shot_outcome"] = df_match["shot"].apply(lambda x: x["outcome"]["name"] if pd.notna(x) and x.get("outcome") else None)
print(df_match["shot_outcome"].value_counts())
```

```python
df_match["team_possession"] = df_match["team"].apply(lambda x: x["name"])
print(df_match["team_possession"].value_counts())
```
